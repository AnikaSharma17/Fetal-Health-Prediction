## Cleaning and splitting

Exact duplicates are dropped. Rows also come in runs that look like consecutive segments of the
same CTG recording, and the CSV has no recording ID, so runs of rows with the same baseline value
are used as a proxy recording ID. The train/test split and every CV split keep each run on one side.

```python
from fetal_health.data import drop_duplicates, label_conflicts

df = drop_duplicates(df)
print("Rows after dropping exact duplicates:", len(df))

# Same features, different label: kept, but worth knowing about
label_conflicts(df)
```

```python
from fetal_health.data import pseudo_recording_groups, split_features
from fetal_health.train import holdout_split

X, y = split_features(df)
groups = pseudo_recording_groups(df)
print(f"{groups.nunique()} pseudo-recordings, median {groups.value_counts().median():.0f} rows, "
      f"largest {groups.value_counts().max()} rows")

X_train, X_test, y_train, y_test, groups_train, groups_test = holdout_split(X, y, groups)
print("Train class distribution:", y_train.value_counts().sort_index().to_dict())
print("Test class distribution:", y_test.value_counts().sort_index().to_dict())
```

```python
from imblearn.over_sampling import SMOTE
import matplotlib.pyplot as plt

from fetal_health.train import RANDOM_STATE

# Illustration only: what SMOTE does to the training set. The models below apply it
# per CV fold through the pipeline, and the test set is never resampled.
_, y_train_resampled = SMOTE(random_state=RANDOM_STATE).fit_resample(X_train, y_train)
print("Training set after SMOTE:", y_train_resampled.value_counts().sort_index().to_dict())

plt.figure(figsize=(4, 4))
y_train_resampled.value_counts().sort_index().plot(kind='bar', color='skyblue')
plt.title('Training set after SMOTE: Normal=1, Suspect=2, Pathologic=3')
plt.xlabel('fetal_health')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.show()
```

## Model selection

Each model is a pipeline of MinMax scaling, SMOTE and the classifier, tuned with grouped,
stratified 5-fold CV on the training set. Models are ranked by macro-F1, which weighs the
three classes equally; accuracy would be dominated by Normal (~78% of rows). The majority-class
baseline shows what a model that ignores the minority classes scores.

```python
from fetal_health.train import search_models

results, best_estimators = search_models(X_train, y_train, groups_train)
results
```

```python
best_model_name = results["model"].iloc[0]
clf = best_estimators[best_model_name]  # already refit on the full training set
clf
```

## Test set evaluation

The held-out test set contains only real records from recordings not seen in training.

```python
from fetal_health.evaluate import per_class_report, summary_metrics

y_pred = clf.predict(X_test)
baseline_pred = best_estimators["Majority baseline"].predict(X_test)

pd.DataFrame({
    best_model_name: summary_metrics(y_test, y_pred),
    "Majority baseline": summary_metrics(y_test, baseline_pred),
}).round(3)
```

```python
per_class_report(y_test, y_pred).round(3)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

from fetal_health.data import CLASS_NAMES

ConfusionMatrixDisplay.from_predictions(y_test, y_pred, display_labels=list(CLASS_NAMES.values()),
                                        cmap='Blues')
plt.title(f'{best_model_name}: test set confusion matrix')
plt.grid(False)
plt.show()
```

## Does grouping matter?

The same tuned pipeline scored with ordinary stratified CV and with grouped CV on the training set.
If stratified CV scores clearly higher, rows from the same recording were leaking across folds and
inflating it.

```python
from fetal_health.evaluate import compare_cv_protocols

compare_cv_protocols(clf, X_train, y_train, groups_train).round(3)
```
