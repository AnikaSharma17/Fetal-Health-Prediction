## Cleaning and splitting

Exact duplicates are dropped. Each row is one segment of a longer CTG recording, and several
segments come from the same recording (often overlapping in time). The Kaggle CSV has no
recording ID, but the original UCI release does, and its rows are in the same order;
`recording_ids.csv` holds those IDs (built by `scripts/build_recording_ids.py`). The train/test
split and every CV split keep all segments of a recording on one side.

```python
from fetal_health.data import drop_duplicates, label_conflicts

RECORDINGS_PATH = Path("recording_ids.csv")  # source recording of each row, from the UCI release
df = load_data(DATA_PATH, RECORDINGS_PATH)  # same data, plus a "recording" column
df = drop_duplicates(df)
print("Rows after dropping exact duplicates:", len(df))

# Same features, different label: kept, but worth knowing about
label_conflicts(df)
```

```python
import numpy as np
from sklearn.metrics import adjusted_rand_score

from fetal_health.data import pseudo_recording_groups, recording_groups, split_features

X, y = split_features(df)
groups = recording_groups(df)
proxy_groups = pseudo_recording_groups(df)  # the heuristic used before the real IDs were recovered

segments = groups.value_counts()
print(f"{groups.nunique()} recordings, median {segments.median():.0f} segments each, largest {segments.max()}")
print("Recordings whose segments carry more than one label:", int((y.groupby(groups).nunique() > 1).sum()))

ids = pd.read_csv(RECORDINGS_PATH)
pairs = overlapping = 0
for _, rec in ids.groupby("recording"):
    start, end = rec["start"].to_numpy(), rec["end"].to_numpy()
    i, j = np.triu_indices(len(rec), k=1)
    pairs += len(i)
    overlapping += int((np.minimum(end[i], end[j]) > np.maximum(start[i], start[j])).sum())
print(f"Segment pairs from the same recording that overlap in time: {overlapping / pairs:.0%}")

print(f"\nOld baseline-run proxy vs real IDs: adjusted Rand index {adjusted_rand_score(groups, proxy_groups):.2f}, "
      f"{int((proxy_groups.groupby(groups).nunique() > 1).sum())} recordings split across proxy groups")
```

```python
from fetal_health.train import holdout_split

X_train, X_test, y_train, y_test, groups_train, groups_test = holdout_split(X, y, groups)
print("Train class distribution:", y_train.value_counts().sort_index().to_dict())
print("Test class distribution:", y_test.value_counts().sort_index().to_dict())
print(f"Test recordings: {groups_test.nunique()} "
      f"(of which with Pathological segments: {groups_test[y_test == 3].nunique()})")
```

```python
from imblearn.over_sampling import SMOTE
import matplotlib.pyplot as plt

from fetal_health.train import RANDOM_STATE

# Illustration only: what SMOTE does to the training set. The models below apply it
# per CV fold through the pipeline, and the test set is never resampled.
_, y_train_resampled = SMOTE(random_state=RANDOM_STATE).fit_resample(X_train, y_train)
print("Training set after SMOTE:", y_train_resampled.value_counts().sort_index().to_dict())

plt.figure(figsize=(4, 4))
y_train_resampled.value_counts().sort_index().plot(kind='bar', color='skyblue')
plt.title('Training set after SMOTE: Normal=1, Suspect=2, Pathologic=3')
plt.xlabel('fetal_health')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.show()
```

## Model selection

Each candidate is a pipeline of MinMax scaling, an imbalance strategy and the classifier, tuned
with grouped, stratified 5-fold CV on the training set. Three imbalance strategies are compared
for every model that supports them:

- **smote**: oversample the minority classes inside each training fold;
- **class_weight**: no resampling, but errors on rare classes cost more (`class_weight="balanced"`);
- **none**: no correction.

Models are ranked by macro-F1, which weighs the three classes equally; accuracy would be dominated
by Normal (~78% of rows). The majority-class baseline shows what a model that ignores the minority
classes scores.

```python
from fetal_health.train import search_models

results, best_estimators = search_models(X_train, y_train, groups_train)
results.round(3)
```

```python
from fetal_health.train import strategy_table

# Resampling ablation: best grouped-CV macro-F1 of each model under each strategy
strategy_table(results).round(3)
```

```python
best_model_name, best_strategy = results.loc[0, ["model", "strategy"]]
clf = best_estimators[best_model_name]  # already refit on the full training set
print(f"Selected: {best_model_name} with {best_strategy}")
clf
```

## Calibrated probabilities and a Pathological threshold

SMOTE and class weights make the model act as if the rare classes were common, so its raw
probabilities may not be trustworthy as risks. Three options compete on out-of-fold log loss within
the training set (grouped by recording): the model's own probabilities ("none"), a sigmoid
calibrator and an isotonic calibrator. Each calibrator is fit on held-out recordings, and the option
with the lowest out-of-fold log loss is kept, so a calibrator is only added when it helps on held-out
training data.

Missing a Pathological case is worse than a false alarm, so instead of taking the most likely
class, the model flags Pathological whenever its probability passes a threshold. The threshold is
the highest one that reaches 95% Pathological recall on out-of-fold training predictions; the test
set plays no part in choosing it.

```python
from fetal_health.decision import choose_calibration, method_name, probability_scores

# Out-of-fold log loss and Brier score of none / sigmoid / isotonic; the lowest log loss wins
calibration_method, oof_scores, oof = choose_calibration(clf, X_train, y_train, groups_train)
print("Calibration method:", method_name(calibration_method))
oof_scores.round(3)
```

```python
from sklearn.calibration import calibration_curve

from fetal_health.decision import fit_probability_model

# The selected model with the chosen calibration (or none), refit on the whole training set
prob_model = fit_probability_model(clf, X_train, y_train, groups_train, calibration_method)
proba_test = prob_model.predict_proba(X_test)

test_proba = {"none": clf.predict_proba(X_test)}
if calibration_method is not None:
    test_proba[method_name(calibration_method)] = proba_test
display(pd.DataFrame({k: probability_scores(y_test, p) for k, p in test_proba.items()}).T.round(3))

fig, axes = plt.subplots(1, 3, figsize=(13, 4))
for col, (label, name) in enumerate(CLASS_NAMES.items()):
    ax = axes[col]
    for kind, p in test_proba.items():
        frac, mean_pred = calibration_curve(y_test == label, p[:, col], n_bins=8, strategy="quantile")
        ax.plot(mean_pred, frac, marker="o", label=kind)
    ax.plot([0, 1], [0, 1], "k--", lw=1)
    ax.set(title=name, xlabel="predicted probability", ylabel="observed fraction")
    ax.legend()
fig.suptitle("Reliability on the test set (closer to the diagonal is better)")
plt.tight_layout()
plt.show()
```

```python
from fetal_health.decision import pathological_threshold, recall_threshold_curve

TARGET_RECALL = 0.95
oof_chosen = oof[calibration_method]  # out-of-fold probabilities with the chosen calibration
threshold = pathological_threshold(y_train, oof_chosen, target_recall=TARGET_RECALL)
print(f"Threshold on P(Pathological): {threshold:.3f}")

curve = recall_threshold_curve(y_train, oof_chosen)
plt.figure(figsize=(5, 3.5))
plt.plot(curve["threshold"], curve["recall"], marker="o", label="recall")
plt.plot(curve["threshold"], curve["precision"], marker="o", label="precision")
plt.axvline(threshold, color="grey", ls="--", label="chosen threshold")
plt.xlabel("threshold on P(Pathological)")
plt.title("Pathological, out-of-fold on the training set")
plt.legend()
plt.show()
```

```python
from fetal_health.decision import predict_with_threshold

y_pred_threshold = predict_with_threshold(proba_test, threshold)
decision_rules = {
    f"{best_model_name} (most likely class)": y_pred,
    f"{method_name(calibration_method)} probabilities (most likely class)": prob_model.predict(X_test),
    "95%-recall Pathological threshold": y_pred_threshold,
}
pd.DataFrame({k: summary_metrics(y_test, p) for k, p in decision_rules.items()}).round(3)
```

```python
threshold_ci = bootstrap_ci(y_test, y_pred_threshold, groups_test)
display(threshold_ci.round(3))

ConfusionMatrixDisplay.from_predictions(y_test, y_pred_threshold,
                                        display_labels=list(CLASS_NAMES.values()), cmap='Blues')
plt.title('95%-recall threshold: test set confusion matrix')
plt.grid(False)
plt.show()
```

## A cost-based decision rule

The 95%-recall threshold serves one goal and ignores Suspect. A more general rule weighs every
kind of error by what it costs and picks two thresholds: flag Pathological if P(Pathological) passes
the first, otherwise Suspect if P(Suspect) passes the second, otherwise Normal. Both thresholds are
chosen together to minimise the mean cost on out-of-fold training predictions (ties go to the
higher thresholds, i.e. fewer alarms); the test set is only used to evaluate the result.

**The costs below (`DEFAULT_COSTS`) are illustrative, not clinical.** They encode one plausible
ordering (any false alarm costs 1, calling a Suspect case Normal 2, calling a Pathological case
Suspect 5 and Normal 10). Real costs would have to come from clinicians and the setting in which the
model is used; the sensitivity table further down shows how much the choice matters.

```python
from fetal_health.decision import (DEFAULT_COSTS, cost_thresholds, costs_with_pathological_miss,
                                   mean_cost, predict_with_thresholds)

class_names = list(CLASS_NAMES.values())
display(pd.DataFrame(DEFAULT_COSTS, index=[f"true {n}" for n in class_names],
                     columns=[f"predicted {n}" for n in class_names]))

cost_path, cost_suspect = cost_thresholds(y_train, oof_chosen)
print(f"Thresholds minimising the out-of-fold mean cost: P(Pathological) >= {cost_path:.2f}, "
      f"then P(Suspect) >= {cost_suspect:.2f}")
y_pred_cost = predict_with_thresholds(proba_test, cost_path, cost_suspect)
```

```python
# The three decision rules on the test set, with 95% recording-level bootstrap intervals
rules = {"most likely class": y_pred, "95%-recall threshold": y_pred_threshold, "cost rule": y_pred_cost}
rule_cis = {name: bootstrap_ci(y_test, p, groups_test) for name, p in rules.items()}
comparison = pd.DataFrame({
    name: ci.apply(lambda r: f"{r['estimate']:.3f} ({r['lower']:.3f}–{r['upper']:.3f})", axis=1)
    for name, ci in rule_cis.items()})
comparison.loc["mean cost (DEFAULT_COSTS)"] = [f"{mean_cost(y_test, p):.3f}" for p in rules.values()]
comparison.loc["Pathological flags"] = [str(int((p == 3).sum())) for p in rules.values()]
comparison.loc["Suspect flags"] = [str(int((p == 2).sum())) for p in rules.values()]
comparison
```

```python
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_cost,
                                        display_labels=list(CLASS_NAMES.values()), cmap='Blues')
plt.title('Cost rule: test set confusion matrix')
plt.grid(False)
plt.show()
```

### How much do the costs matter?

The cost of calling a Pathological case Normal is varied below, with calling it Suspect costing half
as much and the other costs unchanged. For each cost matrix both thresholds are chosen again on the
out-of-fold training predictions only, then applied once to the test set. "Mean cost" is measured
under that row's own cost matrix, so it is not comparable between rows.

```python
METRICS = ["macro_f1", "recall_Pathological", "precision_Pathological", "recall_Suspect",
           "precision_Suspect"]
sensitivity = []
for miss_cost in [2, 5, 10, 20, 50]:
    costs = costs_with_pathological_miss(miss_cost)
    t_path, t_susp = cost_thresholds(y_train, oof_chosen, costs)
    pred = predict_with_thresholds(proba_test, t_path, t_susp)
    sensitivity.append({
        "Pathological miss cost": miss_cost,
        "Pathological threshold": t_path,
        "Suspect threshold": t_susp,
        "OOF mean cost": mean_cost(y_train, predict_with_thresholds(oof_chosen, t_path, t_susp), costs),
        "test mean cost": mean_cost(y_test, pred, costs),
        **{f"test {m}": v for m, v in summary_metrics(y_test, pred).items() if m in METRICS},
        "test Pathological flags": int((pred == 3).sum()),
    })
pd.DataFrame(sensitivity).set_index("Pathological miss cost").round(3)
```

### Why does Suspect precision vary between splits?

`scripts/repeated_splits.py` also saves every test prediction (`results/repeated_predictions.csv`,
one row per test segment and split, with its recording). A false Suspect call is a segment
predicted Suspect whose label is Normal or Pathological. The tables below look at where those
calls come from, for the most likely class unless stated otherwise.

```python
preds = pd.read_csv("results/repeated_predictions.csv")
has_suspect = y.eq(2).groupby(groups).any()  # recording has at least one Suspect-labelled segment
preds["recording_has_suspect"] = preds["recording"].map(has_suspect)


def false_suspect_by_split(preds, rule="default"):
    rows = []
    for seed, split in preds.groupby("seed", sort=False):
        called = split[f"pred_{rule}"] == 2
        false = called & (split["y_true"] != 2)
        per_recording = split[false].groupby("recording").size().sort_values(ascending=False)
        worst = per_recording.index[0] if len(per_recording) else None
        rest = split["recording"] != worst
        rows.append({
            "seed": seed,
            "Suspect calls": int(called.sum()),
            "correct": int((called & ~false).sum()),
            "false": int(false.sum()),
            "precision": (called & ~false).sum() / called.sum(),
            "recordings with false calls": len(per_recording),
            "most from one recording": int(per_recording.max()) if len(per_recording) else 0,
            "precision without that recording": (called & ~false & rest).sum() / (called & rest).sum(),
        })
    return pd.DataFrame(rows).set_index("seed")


by_split = false_suspect_by_split(preds)
display(by_split.round(3))
print(f"Correlation of Suspect precision with the number of false calls: "
      f"{by_split['precision'].corr(by_split['false']):.2f}, with correct calls: "
      f"{by_split['precision'].corr(by_split['correct']):.2f}")
```

```python
# Recordings that produce the false Suspect calls, pooled over the 10 test sets
false_calls = preds[(preds["pred_default"] == 2) & (preds["y_true"] != 2)]
per_recording = (false_calls.groupby("recording")
                 .agg(false_calls=("row", "size"), splits=("seed", "nunique"))
                 .sort_values("false_calls", ascending=False))
per_recording["segments"] = groups.value_counts().reindex(per_recording.index)
per_recording["labels"] = [y[groups == r].value_counts().sort_index().to_dict() for r in per_recording.index]
per_recording["test sets containing it"] = preds.groupby("recording")["seed"].nunique().reindex(per_recording.index)
share = per_recording["false_calls"].cumsum() / len(false_calls)
print(f"{len(false_calls)} false Suspect calls from {len(per_recording)} recordings; "
      f"the top recording gives {share.iloc[0]:.0%}, the top 5 give {share.iloc[4]:.0%}")
per_recording.head(10)
```

```python
# Do false Suspect calls come from recordings that also contain Suspect-labelled segments?
rows = {}
for rule in ["default", "threshold", "cost"]:
    not_suspect = preds[preds["y_true"] != 2]
    false = not_suspect[f"pred_{rule}"] == 2
    rate = false.groupby(not_suspect["recording_has_suspect"]).mean()
    rows[rule] = {
        "false Suspect calls": int(false.sum()),
        "share from recordings with Suspect segments": not_suspect.loc[false, "recording_has_suspect"].mean(),
        "share of non-Suspect test segments in such recordings": not_suspect["recording_has_suspect"].mean(),
        "false-call rate, recordings with Suspect segments": rate[True],
        "false-call rate, recordings without": rate[False],
    }
print(f"{int(has_suspect.sum())} of {groups.nunique()} recordings contain a Suspect-labelled segment")
pd.DataFrame(rows).round(3)
```

What the predictions show:

- **The swings come from false Suspect calls, not from missed ones.** Every test set has 58-59
  Suspect segments and the model calls 39-51 of them correctly, but the number of false Suspect
  calls ranges from 4 to 32. Precision tracks the false calls almost perfectly (correlation -0.98).
- **A few recordings produce most of them.** 139 false Suspect calls over the 10 test sets come from
  32 recordings, and the top 5 give 60%. One recording, S8001041.dsp (17 segments, all labelled
  Normal), alone gives 50 of them (36%): it lands in 5 of the 10 test sets and 9-11 of its segments
  are called Suspect each time. Removing just the worst recording from each test set narrows Suspect
  precision from 0.61-0.92 to 0.69-0.96; seed 8, with false calls spread over 15 recordings, stays
  the lowest.
- **They mostly do not come from mixed-label recordings.** Only 27% of the false Suspect calls come
  from recordings that also contain Suspect-labelled segments. Such recordings do have twice the
  false-call rate per segment (6.7% vs 3.3%), but they hold only 16% of the non-Suspect test segments.
  Most false calls are whole runs of Normal-labelled segments from recordings that were never
  labelled Suspect, which looks like recordings the model finds borderline rather than label noise
  within a recording.

The cost rule calls Suspect more often and has the same pattern (S8001041.dsp again leads). The
95%-recall threshold rule makes fewer Suspect calls (its calibrated probabilities move borderline
segments from Suspect to Normal, and the threshold moves some to Pathological), so it also makes
fewer false ones.
